```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.axes as axes

# 展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

data = pd.read_csv('02.csv',usecols=['电影名字','上映时间','时长'])
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(16, 8))
fig.suptitle('最佳电影年份分布', fontsize=16, y=1)
ax[0].set_title('最佳电影年份分布折线图')
ax[1].set_title('最佳电影年份分布饼状图')

year_count=data.groupby('上映时间')['上映时间'].count()
x = [i for i in range(year_count.index.min(), year_count.index.max()+1)]
y = [year_count.get(i, 0) for i in x]
ax[0].plot(x, y)
ax[0].set_xlabel('年份')
ax[0].set_ylabel('电影数量')
ax[0].grid(True)
ax[0].set_xticks(x[::10])  # 每10年显示一个刻度
y_ticks = [i for i in range(0, max(y)+1)]
ax[0].set_yticks(y_ticks)
                                               
  
```
